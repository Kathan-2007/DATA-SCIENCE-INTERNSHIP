```python
import pandas as pd 
```

```python
df = pd.read_csv("data_cleaning sample.csv")
```

## cleaning data

```python
df1= df[df["Name"].notnull()].sort_values(by='Age',ascending=False).drop_duplicates().fillna(int(df['Age'].mean()))
```

## Export the cleaned dataset

```python
df1.to_csv("Clean_data.csv",index=False)
```
