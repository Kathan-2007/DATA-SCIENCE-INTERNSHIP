# 📊 Jupyter Notebook Mini Dashboard Project

## 📌 Project Overview
This repository contains a mini data analysis dashboard built using Python inside a Jupyter Notebook. The goal of this project was to clean messy data, calculate summary statistics, and generate clear data visualizations (bar chart, pie chart, and histogram).

---

## 📂 Repository Structure
* `notebook.ipynb` - Complete data cleaning and visualization code.
* `data_cleaning sample.csv` - Processed dataset.
* `charts/` - Exported chart images.
* `README.md` - Documentation and submission guide.

---

## 📈 Key Visualizations & Analysis
1. **Bar Chart:** Member breakdown by City and Age analysis.
2. **Pie Chart:** Gender and City combined distribution.
3. **Histogram:** Age distribution across regions.
