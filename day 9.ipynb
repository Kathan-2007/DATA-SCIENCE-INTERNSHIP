```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


# Read dataset
df = pd.read_csv('data_cleaning sample.csv')
print("Initial Dataset:")
display(df)
```

```python
final_dataset=df.drop_duplicates().fillna(0)
final_dataset
```

```python
# Summary Metrics
total_members = len(df)
avg_age = df['Age'].mean()
top_city = df['City'].mode()[0]


print("DASHBOARD SUMMARY")
print("==========================================")
print(f"Total Members : {total_members}")
print(f"Average Age   : {avg_age} years")
print(f"Top Location  : {top_city}")
```

```python

sns.barplot(
    data=final_dataset,
    x="City",
    y="Age",
    hue="City",
    palette="pastel",
    errorbar=None 
)

plt.xlabel("City")
plt.ylabel("Age")
plt.title("CHART OF CITY AND AGE ANALYSIS")
plt.tight_layout()
plt.show()

colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
city_counts = final_dataset['City'].value_counts()
Gender_counts = final_dataset['Gender'].value_counts()
gender_city = final_dataset['Gender'] + ' - ' + final_dataset['City']
counts = gender_city.value_counts()
print("==========================================================================================")
plt.pie(
 counts,labels=counts.index,colors=colors
    
)


plt.title("CHART OF CITY ANALYSIS")
plt.tight_layout()
plt.show()


print("==========================================================================================")

plt.hist(
    final_dataset['City'],  
    bins=5,                 
    color='skyblue',
    edgecolor='black'
)
        
   


plt.xlabel("City")
plt.ylabel("no of employee")
plt.title("CHART OF CITY AND EMPOLYEE ANALYSIS")
plt.tight_layout()
plt.show()
```
